# 4. The tumbling T-handle

Spin an object about its intermediate axis of inertia and it flips over again and again, even with nothing acting on it. This is the Dzhanibekov effect, named after the cosmonaut who saw a wing nut do it in orbit.

Nothing here is scripted: the flips come only from the body's inertia, computed from the CAD shape. There is no gravity and no joint.

```python
import sys
if sys.platform == "emscripten":  # the online version installs its packages here
    %pip install -q lib_rigid_body pythreejs

from netgen.occ import Box, Cylinder, Pnt
from lib_rigid_body import World
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
bar = Box(Pnt(-.6, -.06, -.06), Pnt(.6, .06, .06))
handle = Box(Pnt(-.06, 0, -.06), Pnt(.06, .7, .06))

world = World(gravity=(0, 0, 0))
body = world.add(bar + handle, spin=(0.02, 5, 0.02))  # nearly about the y axis

print("Principal moments of inertia:", np.round(np.diag(body.inertia), 4))
recording = world.run(duration=10, h=0.004)
print(recording.summary())
recording.show()
```

## What flips, and what doesn't

The spin seen from the body reverses again and again, while the world-frame angular momentum and the energy stay constant, as they must without external forces.

```python
spin = recording.spins[:, 0]
momentum = recording.momenta[:, 0]

fig, ax = plt.subplots(figsize=(8, 2.8))
ax.plot(recording.times, spin[:, 1], label=r"$\omega_y$ in the body frame")
ax.set(xlabel="time [s]", ylabel="rad/s")
ax.grid(alpha=.3)
ax.legend()

print("Largest change in angular momentum:", np.max(np.linalg.norm(momentum - momentum[0], axis=1)))
print("Relative energy error:", recording.energy_error)
```

**Try it:** spin about the x axis with `spin=(5, 0.02, 0.02)`. That axis is stable, so the handle won't flip.
