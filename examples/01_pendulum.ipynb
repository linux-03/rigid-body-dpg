# 1. A pendulum modelled in CAD

Model a solid with netgen, pin it to a point, and press play. Mass and inertia come from the geometry.

**Run:** choose *Run > Run All Cells*. The first run takes a little longer because your browser downloads the packages. Everything runs on your own machine.

```python
import sys
if sys.platform == "emscripten":  # the online version installs its packages here
    %pip install -q lib_rigid_body pythreejs

from netgen.occ import Box, Cylinder, Pnt
from lib_rigid_body import World
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
world = World()
cube = world.add(Box(Pnt(-.25, -.25, -.25), Pnt(.25, .25, .25)), at=(1.5, 0, 0), density=100)
world.pin(cube, (0, 0, 0))

recording = world.run(duration=6, h=0.005)
print(recording.summary())
recording.show()
```

## Energy does not drift

A symplectic integrator doesn't conserve energy exactly, but the error oscillates within a band of size $O(h^2)$ instead of growing.

```python
recording.plot_energy();
```

## Heavy and light bodies swing alike

Density scales mass and inertia together, so the motion stays the same. Here the two bodies differ in mass by a factor of 100.

```python
def swing(density):
    world = World()
    cube = world.add(Box(Pnt(-.25, -.25, -.25), Pnt(.25, .25, .25)), at=(1.5, 0, 0), density=density)
    world.pin(cube, (0, 0, 0))
    return world.run(duration=1, h=0.005).positions[-1, 0]

light, heavy = swing(10), swing(1000)
print("light body ends at", light.round(6))
print("heavy body ends at", heavy.round(6))
```

**Try it:** change `at` to start from a different height, or replace the cube with `Cylinder(Pnt(0, 0, -.5), (0, 0, 1), r=.2, h=1)`.
