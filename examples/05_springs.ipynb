# 5. Springs

A spring adds the potential $V = \tfrac{1}{2} k (\lVert d \rVert - L)^2$, where $d$ is the vector between its two ends and $L$ its rest length. We first compare a spring with its exact solution.

```python
import sys
if sys.platform == "emscripten":  # the online version installs its packages here
    %pip install -q lib_rigid_body pythreejs

from netgen.occ import Box, Cylinder, Pnt
from lib_rigid_body import World
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
cube = Box(Pnt(-.1, -.1, -.1), Pnt(.1, .1, .1))
mass, stiffness, amplitude = 2.0, 50.0, 0.3

world = World(gravity=(0, 0, 0))
body = world.add(cube, at=(amplitude, 0, 0), mass=mass, inertia=np.eye(3))
world.spring((0, 0, 0), body, stiffness=stiffness, rest_length=0)
recording = world.run(duration=4, h=0.002)

exact = amplitude * np.cos(np.sqrt(stiffness / mass) * recording.times)
print("Largest difference from the exact solution:", np.max(np.abs(recording.positions[:, 0, 0] - exact)))
```

## A cube on an off-centre spring

Attaching the spring away from the centre of mass couples bouncing, swinging and tumbling. This motion has no simple formula, but the energy error stays small.

```python
world = World()
cube = Box(Pnt(-.2, -.2, -.2), Pnt(.2, .2, .2))
body = world.add(cube, at=(.6, 0, -1.2), density=20)
world.spring((0, 0, 0), body, stiffness=30, at_b=(.15, .15, .2))

recording = world.run(duration=8, h=0.005)
print(recording.summary())
recording.show()
```

```python
recording.plot_energy();
```

**Try it:** change `stiffness`, or attach the spring at the centre with `at_b=(0, 0, 0)`. The cube then stops tumbling, because the spring can no longer apply a torque.
